```python
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser, PydanticOutputParser

str_chain = ChatPromptTemplate.from_template("{topic}(을/를) 한 문장으로 설명해줘") | gpt_model | StrOutputParser()
print(str_chain.invoke({"topic":"머신러닝"}))


prompt = ChatPromptTemplate.from_messages(
    [
        ("system","JSON 형식으로 응답하세요"),
        ("human","{text}의 감정을 분석해서 {{'sentiment':'..','score':0.0}} 형태로 응답"),
    ]
)

json_chain = prompt | gpt_model | JsonOutputParser()
result = json_chain.invoke({"text":"오늘 정말 행복한 하루였어요"})
print(result['sentiment'], result['score'])
```

```python
from pydantic import BaseModel, Field
from typing import Literal

class SentimentResult(BaseModel):
    sentiment:Literal['positive','negative','neutral']
    score:float = Field(ge=0.0, le=1.0, description="감정 강도")
    reason:str = Field(description="판단 근거 한 문장")
    keyword:list[str] = Field(description="핵심 키워드 3개 이내")


pydantic_parser = PydanticOutputParser(pydantic_object=SentimentResult)

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","당신은 감정 분석가입니다. {format_instructions}"),
        ("human","다음 텍스트의 감정을 분석하세요 {text}"),
    ]
).partial(format_instructions=pydantic_parser.get_format_instructions())

chain = prompt | gpt_model | pydantic_parser

response = chain.invoke({'text':'오늘 최악의 하루였습니다. 모든 게 잘못됐어요'})
print(response.sentiment)
print(response.score)
print(response.reason)
print(response.keyword)
```

```python
texts = [
    '이 제품 정말 별로예요. 돈 낭비였습니다.',
    '그냥 평범한 하루였어요. 특별한 건 없었어요',
    '와, 생각보다 훠씬 맛있어요! 강력 추천합니다',
    '제 생각보다는 나았지만 그래도 아쉽네요',
]

# 감정 0.9
# 입력 문장(texts 내 한 줄)
# 근거 이것이 부정/긍정 이라고 판단 한 이유
# 😊😐☹️

response = chain.batch([{'text':t} for t in texts])

print("=== 감정분석 결과 ===")
for text, r in zip(texts,response):
    emoji = {'positive':'😊','negative':'☹️','neutral':'😐'}[r.sentiment]
    print(f'{emoji} [{r.sentiment}] {r.score:.2f}')
    print(f'입력 문장 {text} ')
    print(f'근거 {r.reason}')
    print()
```

```python
# 카테고리 AdvanceSentiment (sentiment,score,category,summary,action)
# 카테고리 => product, service, food, general
# summary => 한줄 요약
# action => 비즈니스 조치 방향

class AdvanceSentiment(BaseModel):
    sentiment:Literal['positive','negative','neutral']
    score:float = Field(ge=0.0, le=1.0, description="감정 강도")
    category:Literal['product','service','food','general'] = Field(description='리뷰 카테고리')
    summary:str = Field(description='한줄 요약')
    action:str = Field(description='비즈니스 조치 방향')

pydantic_parser = PydanticOutputParser(pydantic_object=AdvanceSentiment)

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","당신은 CX(고객경험) 분석가입니다. {format_instructions}"),
        ("human","{text}"),
    ]
).partial(format_instructions=pydantic_parser.get_format_instructions())

chain = prompt | gpt_model | pydantic_parser

response = chain.invoke({'text':'배송은 빠른데 포장이 너무 허술해요'})
print(response.sentiment)
print(response.score)
print(response.category)
print(response.summary)
print(response.action)
```

#### LCEL 심화
- `RunnablePassThrough` : 입력을 그대로 통과시키며 다른 키와 병합할 때 사용
- `RunnableLambda`: 일반 python 함수를 Runnable로 감싸서 체인에 삽입
- `RunnableParallel` : 여러 체인을 동시에 실행하고 결과를 dict 로 합할 때 사용
- `assign()` : 체인 중간에 새 키 추가 가능

```python
from langchain_core.runnables import RunnableLambda

def upper(text):
    return text.upper()

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","당신은 언어 전문가입니다."),
        ("human","{topic} 에 대해 한 문장으로 설명해줘"),
    ]
)

chain = prompt | gpt_model | StrOutputParser() | RunnableLambda(upper)

print(chain.invoke({"topic":"python"}))
```

```python
# '\n' => ' ' 줄바꿈을 공백으로 치환

def remove_newline(text):
    return text.replace("\n"," ")

chain = prompt | gpt_model | StrOutputParser() | RunnableLambda(remove_newline)

print(chain.invoke({"topic":"python"}))
```

```python
from langchain_core.runnables import RunnablePassthrough

prompt =ChatPromptTemplate.from_template('{question}에 대해 한 문장으로 답해줘')
parser = StrOutputParser()

chain = (
    {'question':RunnablePassthrough(),'answer':prompt | gpt_model | parser } 
    | RunnableLambda(lambda x:f"Q {x['question']}\nA: {x['answer']}")
         )

print(chain.invoke("파이썬이란?"))
```

```python
import yfinance as yf
from langchain.agents import create_agent
from langchain_core.tools import tool

@tool
def get_korean_stock_price(code:str):
    """
    한국 주식 종목코드를 입력받아 현재 가격을 조회합니다.

    Args:
    6자리 한국 주식 종목 코드(예: 005930)
    """

    symbol = f"{code}.KS"
    stock = yf.Ticker(symbol)
    data = stock.history(period="1d")

    if data.empty:
        return {"error":"주식 데이터를 찾을 수 없습니다."}

    # 마지막 행 가져오기
    latest = data.iloc[-1]

    return {
        "code":code,
        "current_price":float(latest['Close']),
        "open":float(latest['Open']),
        "High":float(latest['High']),
        "Low":float(latest['Low']),
        "Volume":float(latest['Volume']),
    }

def run_korean_stock_agent(user_input:str):
        
    agent = create_agent(model=gpt_model, tools=[get_korean_stock_price])
    result = agent.invoke({"messages":[{"role":"user","content":user_input}]})
    return result['messages'][-1].text

print(run_korean_stock_agent("005930 주가 알려줘"))        
```

```python
from langchain.agents import create_agent
from langchain_core.tools import tool
from pydantic import BaseModel, Field
from typing import Literal

documents = [
    {
        "title":"AI 기술 보고서",
        "content":"LLM 과 RAG 시스템의 구조를 설명한다."
    },
    {
        "title":"보안 기술 보고서",
        "content":"PQC와 양자내성암호를 설명한다."
    },
    {
        "title":"추천 시스템 보고서",
        "content":"사용자 행동 데이터를 이용한 추천 모델이다."
    },
]

@tool
def search_documents(keyword:str):
    """
    회사 내부 문서를 키워드로 검색한다.
    """
    results = []

    for doc in documents:
        if keyword.lower() in (doc['title'] + doc['content']).lower():
            results.append(doc)

    return results

class SearchDoc(BaseModel):
    summary:str = Field(description="검색 키워드")
    keyword:str = Field(description="검색할 키워드")
    documents:list[str] = Field(description="검색된 문서 제목")

agent = create_agent(model=gpt_model, tools=[search_documents], response_format=SearchDoc)
result = agent.invoke({"messages":[{"role":"user","content":"AI"}]})

print(result['structured_response'])
```

```python
from langchain.agents.middleware import before_model
from langchain.agents import create_agent,AgentState
from langgraph.runtime import Runtime
from langchain.messages import RemoveMessage
from langgraph.graph.message import REMOVE_ALL_MESSAGES

@before_model
def trim_messages(state:AgentState,runtime:Runtime):
    """모델을 호출하기 전에 대화 내역을 첫 메세지를 포함한 최근 3~4개"""
    messages = state['messages']

    # 메세지가 3개 이하면 아무것도 안함
    if len(messages) <=3 :
        return None

    first_msg = messages[0]
    recent_msg = messages[-3:] if len(messages) % 2 == 0 else messages[-4:]
    new_msg = [first_msg] + recent_msg

    return {
        "messages": [
            RemoveMessage(id=REMOVE_ALL_MESSAGES),
            *new_msg
        ]
    }


# middleware ; agent 실행과정에 개입하여 동작을 제어 or 보조
agent = create_agent(model=gpt_model, middleware=[trim_messages], checkpointer=InMemorySaver())
thread_config = {"configurable":{"thread_id":"1"}}

# 대화
agent.invoke(
    {"messages":[{"role":"user","content":"Hi! My Name is Bob."}]}, thread_config
)

agent.invoke(
    {"messages":[{"role":"user","content":"write a short poem about cats"}]}, thread_config
)

agent.invoke(
    {"messages":[{"role":"user","content":"now do the same but for dogs"}]}, thread_config
)

final_response = agent.invoke(
    {"messages":[{"role":"user","content":"What's my name?"}]}, thread_config
)

print(final_response['messages'][-1].content)
```

```python
from langchain.agents.middleware import before_model,after_model
from langchain.agents import create_agent,AgentState
from langgraph.runtime import Runtime
from langchain.messages import RemoveMessage
from langgraph.graph.message import REMOVE_ALL_MESSAGES

@after_model
def delete_messages(state:AgentState,runtime:Runtime):
    """모델을 호출한 후에 오래된 대화 내역을 삭제"""
    messages = state['messages']

    # 앞 대화에서 2개 제거
    if len(messages) > 2 :
        return {
            "messages": [ RemoveMessage(id=m.id) for m in messages[:2]
            ]
        }


# middleware ; agent 실행과정에 개입하여 동작을 제어 or 보조
agent = create_agent(model=gpt_model, middleware=[delete_messages], checkpointer=InMemorySaver())
thread_config = {"configurable":{"thread_id":"1"}}

# 대화
agent.invoke(
    {"messages":[{"role":"user","content":"Hi! My Name is Bob."}]}, thread_config
)

agent.invoke(
    {"messages":[{"role":"user","content":"write a short poem about cats"}]}, thread_config
)

agent.invoke(
    {"messages":[{"role":"user","content":"now do the same but for dogs"}]}, thread_config
)

final_response = agent.invoke(
    {"messages":[{"role":"user","content":"What's my name?"}]}, thread_config
)

print(final_response['messages'][-1].content)
```

```python
# short-term 내용 수정

from langchain.tools import ToolRuntime
from langgraph.types import Command
from langchain_core.messages import ToolMessage

class CustomState(AgentState):
    user_name:str

class CustomContext(BaseModel):
    user_id:str

@tool
def update_user_info(runtime:ToolRuntime[CustomContext,CustomState]):
    """Look up and update user info"""

    user_id = runtime.context.user_id
    name = "John smith" if user_id == "user_123" else "Unknown user"
    return Command(
        update={
            "user_name":name,
            "messages":[
                ToolMessage("Successfully looked up user information",tool_call_id=runtime.tool_call_id)
            ]
        }
    )

@tool
def greet(runtime:ToolRuntime[CustomContext,CustomState]):
    """Use this to greet the user once you found their info"""

    user_name = runtime.state.get('user_name',None)

    if user_name is None:
        return Command(
        update={
            "messages":[
                ToolMessage("update_user_info tool",tool_call_id=runtime.tool_call_id)
            ]
        }
    )
    return f"Hello!!!! {user_name}"



agent = create_agent(model=gpt_model, tools=[update_user_info,greet], state_schema=CustomState , context_schema=CustomState)

result = agent.invoke({"messages":"greet the user"},context=CustomContext(user_id="user_123"))

print(result['messages'][-1].content)
```
