### Simple Prompt 버전
- input : 텍스트 전용 메세지 입력

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model=MODEL,
    input="Write a one-sentence bedtime story about a unicorn.",
)

print(response)
print(response.output_text)
```

```python
from openai import OpenAI

client = OpenAI()

context = [
    {"role":"system","content":"You are a helpful assistant."},
    {"role":"user","content":"파이썬에서 리스트 컴프리헨션이 뭔지 3줄로 응답해 줘"},
]

response = client.responses.create(
    model=MODEL,
    input=context,
)

#print(response)
print(response.output_text)
```

```python
response.output
```

- instructions : 모델의 전반적인 행동 방식이나 응답 원칙을 지정하는 지침(input 에 전달하는 것보다 우선순위가 있음)
- input : 모델에게 전달하는 실제 입력 데이터, 질문, 대화 내용, 문서 등이 포함됨

```python
response = client.responses.create(
    model=MODEL,
    instructions="Always answer in English",
    input=[{"role":"user","content":"한국어로 답변해줘"}],
)
print(response.output_text)
```

```python
# system, developer 

response = client.responses.create(
    model=MODEL,
    input=[
        {"role":"developer","content":"Always answer in English"},
        {"role":"user","content":"한국어로 답변해줘"},
           ],
)
print(response.output_text)
```

```python
response = client.responses.create(
    model=MODEL,
    input="Are semicolons optional in JavaScript?"
)
print(response.output_text)
```

```python
instructions = """
You are a backend API assistant.
- Answer concisely and accurately.
- If unsure, say you are unsure.
- Do not hallucinate APIs or facts.
- Prefer bullet points over long prose.
"""

response = client.responses.create(
    model=MODEL,
    instructions= instructions,
    input="Are semicolons optional in JavaScript?"
)
print(response.output_text)
```

- reasoning : 추론모델의 내부 강도 조절
    - low : 빠르게, 직관적으로
    - medium : 일반적인 설명, 추론
    - high : 복잡한 문제를 더 깊게 추론

```python
response = client.responses.create(
    model=MODEL,
    reasoning={"effort": "low"},
    instructions="Talk like a pirate.",
    input="Are semicolons optional in JavaScript?",
)

print(response.output_text)
```
