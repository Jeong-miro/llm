### Simple Prompt 버전
- input : 텍스트 전용 메세지 입력

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model=MODEL,
    input="Write a one-sentence bedtime story about a unicorn.",
    stream=True
)

print(response)

# 전체 내용 한꺼번에 받을 때
# print(response.output_text)

for event in response:
    print(event)
```

- reasoning : 추론모델의 내부 강도 조절

```python
response = client.responses.create(
    model=MODEL,
    reasoning={"effort": "low"},
    instructions="Talk like a pirate.",
    input="Are semicolons optional in JavaScript?",
)

print(response.output_text)
```

### 멀티턴 대화 (주고받는 대화)
- 이전 대화 내용 유지하기
- stateless : LLM API 는 요청을 독립적으로 처리하기 때문에 이전 대화 내용을 다음 요청에 전달해야 문맥을 이어나갈 수 있다.
- 유지시키는 방법
    1. previous_response_id 이용
    2. 개발자가 문맥 관리
    3. conversations API 이용

```python
# 프랑스의 수도는 어디야? => 파리 => 그 곳의 인구는 얼마나 돼?

context = [{"role":"user","content":"What is the capital of France?"}]

response = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
)

print(response.output_text)
```

```python
# 프랑스의 수도는 어디야? => 파리 => 그 곳의 인구는 얼마나 돼?

context = [{"role":"user","content":"What is the capital of France?"}]

res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
)

# + 붙이기
context += [{"role":"user","content":"What is its populations?"}]

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
    # previous_response_id
    previous_response_id=res1.id
)

print(res1.output_text)
print(res2.output_text)
```

```python
# context 에 값을 넣기

context = [{"role":"user","content":"What is the capital of France?"}]

res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
)

# + 붙이기

context += res1.output
context += [{"role":"user","content":"What is its populations?"}]

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
)

print(res1.output_text)
print(res2.output_text)
```

```python
# previous_response_id 로 값을 넣기

context = [{"role":"user","content":"What is the capital of France?"}]

res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
    store=True # default True 긴 함
)

# + 붙이기

context = [{"role":"user","content":"What is its populations?"}]

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
    previous_response_id=res1.id
)

print(res1.output_text)
print(res2.output_text)
```

```python
context = [
    {"role":"user","content":"What is the capital of France?"},
    {"role":"assistant","content":"Paris"},
    {"role":"user","content":"What is its populations?"},
        
    ]

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=context,
)
print(res2.output_text)
```

- Conversation API
    - 대화 객체로 관리
    - 여러 세션이나 작업에 걸쳐 대화를 이어나갈 수 있다.

```python
# conversation API 이용하기
import openai

# conversation 객체 생성
conversation = openai.conversations.create()

res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=[{"role":"user","content":"농담 하나 해줘"}],
    conversation=conversation.id,
)

res2 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=[{"role":"user","content":"그게 왜 재미있는지 이유 설명해줘"}],
    conversation=conversation.id,
)

print(res1.output_text)
print(res2.output_text)
```

#### 파라메터

```python
# Unsupported parameter: 'temperature" is not supported with this model.
# (reasoning 모델에서는 reasoning 을 none 안할경우)
res1 = client.responses.create(
    model=MODEL,
    reasoning={"effort":"none"},
    input=[{"role":"user","content":"Tell me a story"}],
    temperature=1.5
)

print(res1.output_text)
```
